```python
import pandas as pd
df = pd.read_csv('train.csv')

#Основная информация о датасете
df.info()
df.describe()
df.isnull().sum()

#Процент выживаемости у каждого класса
df.groupby(["Pclass"])["Survived"].mean()*100

#Самое популярное мужское и женское имя на корабле
pop1 = df[df['Sex'] == 'female']['Name'].value_counts().index[0]
print(f"Самое популярное женское имя: {pop1}")
pop2 = df[df['Sex'] == 'male']['Name'].value_counts().index[0]
print(f"Самое популярное мужское имя: {pop2}")

#Самое популярное мужское и женское имя в каждом классе
df.groupby(['Pclass', 'Sex'])['Name'].agg(lambda x: x.value_counts().index[0])

#Пассажиры младше 44 лет
df[df['Age']>44]

#Пассажиры - мужчины младше 44 лет
df[(df['Age']<44)&(df['Sex']=='male')]

#Количество n-местных кабин
df['Cabin'].value_counts().value_counts()
```
